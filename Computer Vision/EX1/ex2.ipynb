```python
import numpy as np

# 打开IN文件
with open('C:/Users/15541/Desktop/2sem/geo/progam2/5.in', 'r') as f:
    # 读取m和n
    m = int(f.readline().strip())
    n = int(f.readline().strip())

    # 读取n个点的名称和坐标
    data = []
    for i in range(n):
        line = f.readline().split()
        name = line[0]
        coords = np.array([float(x) for x in line[1:]])
        data.append((name, coords))

    # 创建n*m的numpy array
    mul = np.zeros((1, n))
    point = np.zeros((1, m))
    n_array = np.empty((n, m))
    for i in range(n):
        n_array[i] = data[i][1]
        
    # 读取p个点的名称和坐标
    p = int(f.readline().strip())

    # 构造affine约束矩阵
    a_n = np.zeros((1, n+p))
    a_n[:, :n] = 1
    a_p = np.zeros((1, n+p))
    a_p [0, -p:] = 1
    a = np.zeros((m+2, 1))
    a[-2:] = 1

    queries = []
    for i in range(p):
        line = f.readline().split()
        name = line[0]
        coords = np.array([float(x) for x in line[1:]])
        queries.append((name, coords))

    # 创建p*m的numpy array
    p_array = np.empty((p, m))
    for i in range(p):
        p_array[i] = queries[i][1]

    # 整合成矩阵
    nep_array= -p_array
    ma = np.concatenate((n_array, -p_array), axis=0)
    ma0 = np.transpose(ma)
    ma1 = np.concatenate((ma0, a_n), axis=0)
    matrix = np.concatenate((ma1, a_p), axis=0)
    all_matrix = np.concatenate((matrix, a), axis=1)


    if np.linalg.matrix_rank(matrix) != np.linalg.matrix_rank(all_matrix):
        print("N")
    else:
            # 求解方程组
        arr = np.zeros(m + 2)
        arr[-2:] = 1
        x, residuals, rank, s = np.linalg.lstsq(matrix, arr,rcond=None)
        for i in range(n):
            mul[0][i] = x[i]
        end = np.dot(mul, n_array)
        output = "Y"
        for i in range(m):
            output += " "
            output += str(end[0][i])
        print(output)
```
